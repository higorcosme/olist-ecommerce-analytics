```python
import pandas as pd
from sqlalchemy import create_engine

# Configura conexão com SQL Server
server = 'localhost'
database = 'OlistDB'

## Define a string de conexão com o banco de dados
connection_string = f"mssql+pyodbc://@{server}/{database}?driver=ODBC+17+Driver+for+SQL+Server&trusted_connection=yes"

## Cria a engine do SQLAlchemy, responsável por gerenciar conexão com o banco
engine_olist = create_engine(connection_string)

print("Conexão criada com sucesso!")
```

```python
## Importando o csv de pedidos
df_orders = pd.read_csv("..\data\olist_orders_dataset.csv")

## Lista para guardar nome das colunas que serão convertidas
cols_data = [
    "order_purchase_timestamp",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

## Altera o tipo das colunas para datetime
for col in cols_data:
    df_orders[col] = pd.to_datetime(df_orders[col])

## Adiciona a tabela de pedidos no SQL Server
df_orders.to_sql('pedidos', con=engine_olist, if_exists='replace', index=False, chunksize=10000)

print("Tabela de pedidos adicionada com sucesso!")
```

```python
## Importando o csv de produtos
df_products = pd.read_csv("..\data\olist_products_dataset.csv")

## Adiciona a tabela de produtos no SQL Server
df_products.to_sql('produtos', con=engine_olist, if_exists='replace', index=False, chunksize=10000)

print("Tabela de produtos adicionada com sucesso!")
```

```python
## Importando o csv de clientes
df_customers = pd.read_csv("..\data\olist_customers_dataset.csv")

## Adiciona a tabela de clientes no SQL Server
df_customers.to_sql('clientes', con=engine_olist, if_exists='replace', index=False, chunksize=10000)

print("Tabela de clientes adicionada com sucesso!")
```

```python
## Importando o csv de itens
df_items = pd.read_csv("..\data\olist_order_items_dataset.csv")

## Convertendo tipo para datetime
df_items['shipping_limit_date'] = pd.to_datetime(df_items['shipping_limit_date'])

## Adiciona a tabela de intos no SQL Server
df_items.to_sql('itens', con=engine_olist, if_exists='replace', index=False, chunksize=10000)

print("Tabela de itens adicionada com sucesso!")
```
